```python
import pandas as pd
from sklearn.linear_model import LinearRegression
data = {
    "Years_experience" :[1, 2, 3, 4, 5, 6, 7],
    "Salary" :[20, 40, 60, 80, 100, 140, 160]
}
df = pd.DataFrame(data)
print(df)
X = df["Years_experience"] #X contains the input
y = df["Salary"] #Y contains the output
model = LinearRegression() #creating the model
model.fit(X.to_frame(), y) #Training the model
prediction = model.predict([[9]])
print(prediction)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y)
plt.plot(X, model.predict(X.to_frame()), color="red")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
data = { "temp":[20, 22, 23, 24, 26, 28],
        "IceCreamSales":[200, 205, 250, 260, 300, 310]}
df = pd.DataFrame(data)
print(df)
X = df["temp"]
y = df["IceCreamSales"]
model = LinearRegression().fit(X.to_frame(),y)
print("Prediction:", model.predict([[25]]) [0])
print("coefficient:", model.coef_[0])
print("intercept:", model.intercept_)
plt.scatter(X, y)
plt.plot(X, model.predict(X.to_frame()), color="red")
plt.xlabel("Temperature")
plt.ylabel("IceCreamSales")
plt.show()
```
