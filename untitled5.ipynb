```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
Students = pd.DataFrame({
    "Student":["John", "Mary", "Ali", "Grace",
               "James", "Faith", "Brian", "Lucy"],
    "Age":[20,21,22,20,23,21,22,20],
    "Gender":["Male","Female","Male","Female",
              "Male","Female","Male","Female"],
    "Math":[78,85,90,66,74,88,90,76],
    "Science":[80,79,95,70,88,90,76,94],
    "English":[75,88,92,72,81,86,79,90]
})
```

```python
print(Students)
```

```python
#A line chart shows trend over time or ordered data.
plt.plot(Students["Student"], Students["Math"])
plt.show()
```

```python
#Adding titles
plt.plot(Students["Student"], Students["Math"])
plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
students = ["A","B","C","D","E"]
marks = [60,75,56,89,70]
plt.plot(students, marks)
plt.title("Science Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.bar(Students["Student"], Students["Math"])
plt.show()
```

```python
import matplotlib.pyplot as plt
learners = pd.DataFrame({
    "Name":["John", "Mary", "Ali", "Grace",
               "James", "Faith", "Brian", "Lucy"],
    "Maths":[78,21,56,20,65,91,22,76],
    "Science":[80,79,95,70,88,90,76,94],
    "English":[75,88,92,72,81,86,79,90]
})
print(learners.head())
learners.plot(x="Name", y=["Maths", "Science", "English"], kind="bar")
plt.show()
learners["English"].plot(kind="hist")
plt.show()
plt.scatter(learners["Maths"], learners["Science"])
plt.xlabel("Math")
plt.ylabel("Science")
plt.show()
```

saeborn

```python
import seaborn as sns
```

```python
sns.barplot(data=Students, x="Gender", y="Math")
plt.show()
```

```python
#Scater plot
sns.scatterplot(data=Students, x="Math", y="Science", hue="Gender")
plt.show()
```

```python
#Heatmap - Displays correlations between numerical variables
corr = Students[["Math", "Science", "English"]].corr()
sns.heatmap(corr, annot=True, cmap="Blues")
plt.show()
```
