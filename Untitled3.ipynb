```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.utils import resample
```

3. CLEAN THE DATA.

```python
df = df.drop(columns=['id'])                      # identifier, no predictive value
df = df[df['gender'] != 'Other'].copy()           # single record

# Impute missing BMI with the median of each age group
age_group = pd.cut(df['age'], [0, 18, 40, 60, 120])
df['bmi'] = df.groupby(age_group, observed=False)['bmi'].transform(lambda s: s.fillna(s.median()))

# Winsorise extreme BMI values at the 1st and 99th percentiles
low, high = df['bmi'].quantile([0.01, 0.99])
df['bmi'] = df['bmi'].clip(low, high)

print("Missing values after cleaning:", df.isnull().sum().sum())
print(df['bmi'].describe().round(2))
```

 4. ENCODE CATEGORICAL VARIABLES.

```python
df['ever_married'] = df['ever_married'].map({'No': 0, 'Yes': 1})
df['Residence_type'] = df['Residence_type'].map({'Rural': 0, 'Urban': 1})
df = pd.get_dummies(df, columns=['gender', 'work_type', 'smoking_status'], drop_first=True, dtype=int)
print(df.shape)
print(list(df.columns))
df.head()
```

5. SPLIT INTO TRAINING AND TESTING SETS,THEN SCALE.

```python
X = df.drop(columns='stroke')
y = df['stroke']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y)
print("Train:", X_train.shape, " Test:", X_test.shape)
print("Stroke rate train: %.2f%%  test: %.2f%%" % (y_train.mean()*100, y_test.mean()*100))
```

FEATURE SCALING

```python
num_cols = ["age","avg_glucose_level","bmi"]
print("BEFORE scaling (train):")
print(X_train[num_cols].describe().loc[["mean", "std"]]. round(2))

scaler = StandardScaler ()
X_train[num_cols] = scaler.fit_transform(X_train [num_cols])
X_test[num_cols] = scaler.transform(X_test[num_cols])
print ("\nAFTER scaling (train):")
print (X_train[num_cols].describe().loc[["mean", "std"]].round(2))
```

6. HANDLE CLASS IMBALANCE(M2)

```python
train = X_train.copy(); train['stroke'] = y_train
majority, minority = train[train.stroke == 0], train[train.stroke == 1]
minority_up = resample(minority, replace=True, n_samples=len(majority), random_state=42)
balanced = pd.concat([majority, minority_up])
X_bal, y_bal = balanced.drop(columns='stroke'), balanced['stroke']

print("Before:", y_train.value_counts().to_dict())
print("After :", y_bal.value_counts().to_dict())
```

```python
# AFTER charts
fig, ax = plt.subplots(1, 3, figsize=(12, 3.6))
ax[0].bar(["Before", "After"], [raw.bmi.isna().sum(), df.isnull().sum().sum()], color=["#c0392b", "#27ae60"])
ax[0].set_title("Missing values")
ax[1].hist(raw.bmi.dropna(), bins=40, alpha=.55, label="Before", color="#c0392b")
ax[1].hist(raw.bmi.dropna().clip(low, high), bins=40, alpha=.55, label="After (capped)", color="#2e75b6")
ax[1].legend(); ax[1].set_title("BMI distribution")
w = .35; b = y_train.value_counts().sort_index().values; a = y_bal.value_counts().sort_index().values
x = np.arange(2)
ax[2].bar(x - w/2, b, w, label="Before", color="#c0392b"); ax[2].bar(x + w/2, a, w, label="After", color="#27ae60")
ax[2].set_xticks(x); ax[2].set_xticklabels(["No stroke", "Stroke"]); ax[2].legend(); ax[2].set_title("Training class balance")
plt.tight_layout(); plt.show()
```

7. BEFORE AND AFTER; DOES THE REFINED DATA IMPROVE THE MODEL

```python
def evaluate(Xtr, ytr, Xte, yte):
    pred = LogisticRegression(max_iter=1000).fit(Xtr, ytr).predict(Xte)
    return dict(Accuracy=accuracy_score(yte, pred), Precision=precision_score(yte, pred, zero_division=0),
                Recall=recall_score(yte, pred), F1=f1_score(yte, pred, zero_division=0))

# BEFORE: minimal preparation (median BMI, one-hot, no capping, no scaling, no balancing)
r = raw.drop(columns=['id']).copy()
r['bmi'] = r['bmi'].fillna(r['bmi'].median())
r = pd.get_dummies(r, drop_first=True).astype(float)
Xr, yr = r.drop(columns='stroke'), r['stroke']
Xr_tr, Xr_te, yr_tr, yr_te = train_test_split(Xr, yr, test_size=.2, random_state=42, stratify=yr)
before = evaluate(Xr_tr, yr_tr, Xr_te, yr_te)

# AFTER: refined pipeline
after = evaluate(X_bal, y_bal, X_test, y_test)

pd.DataFrame({'Before refinement': before, 'After refinement': after}).round(3)
```
