```python
import pandas as pd
```

```python
sample_data = {
"Name": ["Salma", "Babu", "Nand", "Irfan"],
"Math": [78, 34, 29, 33],
"English":[58, 74, 69, 93],
"Science":[68, 54, 59, 91],
"History":[78, 64, 50, 41],
}
Students = pd.DataFrame(sample_data)
Students.to_csv("students.csv", index = False)
new_data = pd.read_csv("students.csv")
print(new_data)
```

**viewing data:** pandas provide functions for viewing and inspecting data set

```python
print(new_data.head()) #Display the first 5 rows
print(new_data.head(10)) #Display the first 10 rows
print(new_data.tail()) #Display the last rows
print(new_data.info()) #Display dataset info:
print(new_data.shape) #Display the shape of the data
print(new_data.describe()) #Display the statistical summary
print(new_data.columns) #show column names
```
