```python
import pandas as pd
```

```python
sample_data = {
"Name": ["Salma", "Babu", "Nand", "Irfan"],
"Math": [78, 34, 29, 33],
"English":[58, 74, 69, 93],
"Science":[68, 54, 59, 91],
"History":[78, 64, 50, 41],
}
Students = pd.DataFrame(sample_data)
Students.to_csv("students.csv", index = False)
new_data = pd.read_csv("students.csv")
print(new_data)
```

```python
print(Students["Math"]) #Selecting one column
print(Students[["Name", "Math"]]) #Selecting multiple column
Math_scores = Students["Math"] #Saving selected columns
print(Math_scores)
```

Selcting Rows: (ioc[0] select rows using labels)

```python
print(Students.loc[0]) #selecting one row
print(Students.loc[[0, 1,]]) #selecting multiple columns
#iloc[] selects rows using position numbers
print(Students.iloc[0]) #selecting one row
print(Students.iloc[[0, 1]]) #selecting multiple rows
```

```python
print(Students.loc[0, "Name"]) #Selecting one row and one column
print(Students.iloc[[0, 1], [0, 1]]) #multiple rows and columns by position
```

```python
#Filtering data
#students with more than 60 in maths
highscore = Students[Students["Math"]>60]
print(highscore)
print(Students["Math"].mean()) #average score in maths
print(Students["Math"].min())  #minimum score in maths
print(Students["Math"].max())  #maximum scores in maths
print(Students["Math"].sum())  #total score in maths
```

```python
import pandas as pd
data = {
    "track_name": ["Blinding Lights", "Shape of You", "Perfect", "Believer"],
    "artist": ["The Weeknd", "Ed Sheeran", "Ed Sheeran", "Dua Lipa"],
    "genre": ["Pop", "Pop", "Pop", "Pop"],
    "duration_mins": [3.20, 4.10, 2.50, 3.80],
    "popularity_score": [95, 92, 88, 85]
}
spotify = pd.DataFrame(data)
print(spotify.head(1))
print(spotify["popularity_score"]==spotify["popularity_score"].max())
print(spotify["duration_mins"].mean())
print(spotify[spotify["popularity_score"]>70])
```
