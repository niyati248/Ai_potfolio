#**Data frames**
a data frame is a 2 dimensional table. It contains rows and colimns. Similar to an excel spreadsheets

```python
data = {
    "Name": ["Salma", "Babu", "Nand", "Irfan"],
    "Math": [78, 34, 29, 33],
    "English": [58, 74, 69, 93],
    "Science": [68, 54, 59, 93]

}
students = pd.DataFrame(data)
print(students)
```

```python
print(students.dtypes) # Checking data types
print(students.shape) # Checking the shape of the data frame
```
